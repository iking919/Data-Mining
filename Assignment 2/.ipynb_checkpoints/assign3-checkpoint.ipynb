```python
#Inputs: P: list of tuples (itemset, tidset), minsup: minimum support (integer), C: set of closed itemsets (empty)

def CHARM(P, minsup, C):
    P = sorted(P, key=lambda x: len(x[1]))    # Sort P by support (length of tidset)
    i = 0
    while i < len(P):
        Xi, tXi = P[i]
        Pi = []
        j = i + 1
        
        while j < len(P):
            Xj, tXj = P[j]
            Xij = tuple(sorted(set(Xi) | set(Xj)))     # Use | for set union, sort for output comparison later
            tXij = tXi & tXj                           # Use & for set intersection
            
            if len(tXij) >= minsup:
                if tXi == tXj:                         # Property 1
                    P[i] = (Xij, tXi)                  # Replace Xi with Xij in P
                    P.pop(j)                           # Remove (Xj, tXj) from P
                    continue                           # Don't increment j, because list shifted
                    
                elif tXi < tXj:                # Property 2, use < for subset
                    P[i] = (Xij, tXi)          # Replace Xi with Xij in P
                    
                
                else:                             # Property 3
                    Pi.append((Xij, tXij))
            j += 1
        
        
        if Pi:
            CHARM(Pi, minsup, C)
        
        doesNotExist = True            # doesNotExist variable will track if there exists a Z such that Xi subseteq Z and t(Xi) = t(Z)
        for Z, tZ in C:
            if set(Xi).issubset(Z) and tXi == tZ:
                doesNotExist = False
                break
        if doesNotExist:             # Add Xi to closed set if no superset with same tidset exists
            C.append((Xi, tXi))
        
        i += 1
```

```python
# Read the transactions from a text file
transactions = []
with open("ModApte_words.txt", "r") as f:
    for i, line in enumerate(f):
        # Split each line into words and convert to a set
        words = set(line.strip().split())
        transactions.append((words, {i}))  # Each transaction has its tidset as {i}


minsup = 50
C = []

CHARM(transactions, minsup, C)
print(f"Frequent Closed Itemsets with minsup = {minsup}:\n")
for itemset, tidset in C:
    print(f"Itemset: {itemset}, Support: {len(tidset)}")
```
