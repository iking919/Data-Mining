## Data Parsing

We start by using Pandas Library to parse our data file, then we use NumPy to create our Data Matrix

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

rawData = pd.read_csv("wdbc.data")          #Use Pandas Library to parse inital data file, store in Data frame
diagnosisData = (rawData.iloc[ : , 1]).to_numpy()          #The Second column is the Diagosis Variable, which we will use for plotting
dataMatrix = (rawData.iloc[ : , 2: ]).to_numpy()           #Columns 3-32 are saved and added created into a Data Matrix using NumPy. Column 1, Patient ID, is disregarded.
```

## Scale and Center the Data Matrix
To accomplish this, we use sklearn's MinMaxScaler to create a scaler used to scale all our data to be between 0 and 1. Next we use NumPy to calculate the mean vector for each attribute, which we use to subtract and center our data about the origin. We are left with fixedDataMatrix, which is our scaled and centered data matrix D.

```python
scaler = MinMaxScaler(feature_range = (0,1))          #Use sklearn's MinMaxScaler to fit our data between (0,1)
scaledDataMatrix = scaler.fit_transform(dataMatrix)   

meanValues = np.mean(scaledDataMatrix, axis = 0)      #Use NumPy to get means for each column in our matrix
fixedDataMatrix = scaledDataMatrix - meanValues       #Use the means we just calculated to subtract from our data, centering it about the origin.
```

## Compute Total Variance
Since our data matrix is already centered, we know the mean vector of each column is 0. Thus, we wont have to compute and subtract the mean vectors when computing the Total Variance. So, instead, we can calculate this by summing up the squared norm of each row in our matrix. We get the Total Variance, totVar(D) $\approx 351.8433$

```python
totVar = np.sum(np.linalg.norm(fixedDataMatrix, axis=1)**2)                       #Use NumPy to sum the squared norms of our Matrix, giving us our Total Variance

print(f"Total Variance, totVar(D) = {totVar}")
```

## Find Best Projection Vector
We use NumPy to generate random unit vectors. From this we calculate it's mean squared error based on the our data matrix from before. We go through 10,000 random vectors, storing which vector has the smallest MSE value. This will serve as a our "Best" projection vector based on this random approach.

```python
u1 = None          #Create u1 and minMSE, which will keep track of our random vector that yields the least mean squared error
minMSE = None
for i in range(10000):           #We will go through this process 10000 times
    randVector = np.random.randn(fixedDataMatrix.shape[1])      #Use NumPy to find a random vector of d elements
    unitVector = randVector / np.linalg.norm(randVector)        #We find the unit vector of the previous random vector
    MSE = 0
    
    for x_i in fixedDataMatrix:                       #Go through each row of our matrix, calculating it's projection onto our random unit vector, and then using it to calculate the mean squared error
        p_i = np.dot(x_i, unitVector) * unitVector
        MSE += np.linalg.norm(x_i - p_i)**2
        
    if i == 0 or MSE < minMSE:                        #If this is our first iteration, or best iteration, we store the values for the unit vector and it's MSE, allowing us to track the best vector.
        u1 = unitVector
        minMSE = MSE
        
print(f"Smallest MSE value obtained for u1: {minMSE}")
```

## Find Second Best Vector
Following a similar approach, we now use NumPy to find random vectors that are orthogonal to our last best Projection Vector. We do this another 10,000 times, so we now have our best projection vector, and the best respective orthogonal vector we could find based on this random approach.

```python
u2 = None         #Similarly, we create u2 and minMSE2 which will keep track for the orthogonal vectors
minMSE2 = None
for i in range(10000):          #We will go through this process 100000 times
    randVector = np.random.randn(fixedDataMatrix.shape[1])         #We use NumPy to find a random vector of d elements
    p_2 = np.dot(randVector, u1) * u1                              #We make that random vector orthogonal to u1
    orthoVector = randVector - p_2 
    unitVector = orthoVector / np.linalg.norm(orthoVector)         #Now we go through the same process of comparing MSE to find the "best" orthogonal vector we generate.
    MSE = 0
    
    for x_i in fixedDataMatrix:
        p_i = np.dot(x_i, unitVector) * unitVector
        MSE += np.linalg.norm(x_i - p_i)**2
        
    if i == 0 or MSE < minMSE2:
        u2 = unitVector
        minMSE2 = MSE
        
print(f"Smallest MSE value obtained for u2: {minMSE2}")
```

## Project Data and Plot
Now we have u1 and u2, and we project the centered data matrix onto each of these vectors. From this n x 2 dataset we plot this, with the Benign samples being Green and the Malignant samples being Red. Lastly we print the fraction of total variance captured by these two dimensions. We calculate this by summing up the projected variances and dividing by the entire total variance we calculated before.

```python
proj_u1 = np.dot(fixedDataMatrix, u1)      #The dot product gives us the projection of our data on u1
proj_u2 = np.dot(fixedDataMatrix, u2)      #Likewise, we get the projection on u2

projDataSet = [proj_u1, proj_u2]           #We now get our entire projected data set


colors_list = []                           #Now, we loop through each of our samples, using our diagnosis data from before to label it red if it's malignant or green if it's benign
for d in diagnosisData:
    if d == 'M':
        colors_list.append('red')
    else:
        colors_list.append('green')
colors = np.array(colors_list)             #Once we have our list of colors we can make it into an array to plot

plt.xlabel("Projection on u1")             #We label our plot title and axis
plt.ylabel("Projection on u2")
plt.title("Projection of Data onto u1 and u2")
plt.scatter(projDataSet[0], projDataSet[1], c = colors)       #Create scatter plot with our projected data and colors
plt.show()        

var_u1 = np.sum((proj_u1)**2)           #Calculate total variance from u1 projection
var_u2 = np.sum((proj_u2)**2)           #Calculate total variance from u2 projection
projected_var = var_u1 + var_u2         #We get our total variance for our projection

varFraction = projected_var / totVar    #Lastly we compare this to our entire total variance to get the fraction of total variance captured
print(f"Fraction of total variance captured: {varFraction}")
```

## Improving Directions

We use the idea of a local search to generate new vectors that are slightly in a new direction, rather than completely random. We still generate random vectors, but use a perturb value so that the values are on a much smaller scale, i.e. 0.01. By doing this we can search "around" the previosu best direction. We start with the previously generated u1, and go through 1000 searches. Next, we do this again using a new best direction if we found one. We repeat this 10 times. Likewise, we generate a random vector u2 that is orthogonal to u1. From there we do the same local search 1000 times for 10 iterations. Then, we plot our data the same way we did  before. Based on our plot and our fraction of total variance captured, we can conclude that the local search was more effective.

```python
uCurr = u1               #We start our local search with the best random vector we found, and we know its MSE
MSECurr = minMSE
u3 = None        
perturb = 0.01          #This is our perturb value which we can use scale how small we want to adjust our vector u1


#The following is a nested for loop. In each iteration, it takes our current best vector and will use our perturb value to find local vectors within that value away. If this new vector has a smaller
#MSE, and is a better fit, we will store it. After 1000 iterations, we restart the process, using the best new vector as the "mean" which we will search around. This is done 10 times.
for i in range(10):      
    minMSE3 = None
    
    for j in range(1000):          
        randVector = np.random.randn(fixedDataMatrix.shape[1]) * perturb     
        adjustedVector = uCurr + randVector     
        unitVector = adjustedVector / np.linalg.norm(adjustedVector)         
        MSE = 0
        
        for x_i in fixedDataMatrix:
            p_i = np.dot(x_i, unitVector) * unitVector
            MSE += np.linalg.norm(x_i - p_i)**2
            
        if j == 0 or MSE < minMSE3:
            u3 = unitVector
            minMSE3 = MSE

    if minMSE3 < MSECurr:
        uCurr = u3
        MSECurr = minMSE3

u1 = uCurr         #We store the new best vector as u1, and the first minMSE
minMSE = MSECurr


#Now, we will do the same random local search for u2. We have to first start with the initial random orthogonal vector. We compute that here, befoer beginning our search looop

uCurr = None        
MSECurr = None
for i in range(10000):          #We will go through this process 100000 times
    randVector = np.random.randn(fixedDataMatrix.shape[1])         #We use NumPy to find a random vector of d elements
    p_2 = np.dot(randVector, u1) * u1                              #We make that random vector orthogonal to u1
    orthoVector = randVector - p_2 
    unitVector = orthoVector / np.linalg.norm(orthoVector)         #Now we go through the same process of comparing MSE to find the "best" orthogonal vector we generate.
    MSE = 0
    
    for x_i in fixedDataMatrix:
        p_i = np.dot(x_i, unitVector) * unitVector
        MSE += np.linalg.norm(x_i - p_i)**2
        
    if i == 0 or MSE < MSECurr:
        uCurr = unitVector
        MSECurr = MSE

#Since we have our initial u2 vector, we can go through the same process we did for u1, but now the new shifted vectors we find we have to make orthogonal to u1. We do this 1000 times, over 10 iterations
for i in range(10):
    minMSE3 = None
    
    for j in range(1000):          
        randVector = np.random.randn(fixedDataMatrix.shape[1]) * perturb     
        adjustedVector = uCurr + randVector     
        p_2 = np.dot(adjustedVector, u1) * u1
        orthoVector = adjustedVector - p_2
        unitVector = orthoVector / np.linalg.norm(orthoVector)
        MSE = 0
        
        for x_i in fixedDataMatrix:
            p_i = np.dot(x_i, unitVector) * unitVector
            MSE += np.linalg.norm(x_i - p_i)**2
        
        if j == 0 or MSE < minMSE3:
            u3 = unitVector
            minMSE3 = MSE

    # Update u2 if improvement found
    if minMSE3 < MSECurr:
        uCurr = u3
        MSECurr = minMSE3
u2 = uCurr
minMSE2 = MSECurr


#Now with our new vectors u1 and u2 we can follow the same process as before to plot these points
 
proj_u1 = np.dot(fixedDataMatrix, u1)      
proj_u2 = np.dot(fixedDataMatrix, u2)      

projDataSet = [proj_u1, proj_u2]          


colors_list = []                           
for d in diagnosisData:
    if d == 'M':
        colors_list.append('red')
    else:
        colors_list.append('green')
colors = np.array(colors_list)           

plt.xlabel("Projection on u1")            
plt.ylabel("Projection on u2")
plt.title("Projection of Data onto u1 and u2")
plt.scatter(projDataSet[0], projDataSet[1], c = colors)       
plt.show()        

var_u1 = np.sum((proj_u1)**2)          
var_u2 = np.sum((proj_u2)**2)          
projected_var = var_u1 + var_u2        

varFraction = projected_var / totVar    
print(f"Fraction of total variance captured: {varFraction}")

```
